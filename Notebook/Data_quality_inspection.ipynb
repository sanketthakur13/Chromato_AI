```python
import pandas as pd
from pathlib import Path

data_dir = Path.cwd().parent / "data" / "raw"

df1 = pd.read_excel(data_dir / "dataset_1013.xlsx")
df2 = pd.read_csv(data_dir / "rf_dataset_0203_total.csv")
df3 = pd.read_excel(data_dir / "TLC_dataset.xlsx")

print("All datasets loaded successfully!")
```

```python
datasets = {
    "Column Chromatography": df1,
    "TLC Dataset 1": df2,
    "TLC Dataset 2": df3
}

for name, df in datasets.items():
    print(f"\n{'='*50}")
    print(name)
    print("Rows and columns:", df.shape)
    print("\nMissing values:")
    print(df.isnull().sum())
    print("\nDuplicate rows:", df.duplicated().sum())
    print("\nData types:")
    print(df.dtypes)
```

```python
print("Dataset 1 solvent systems:")
print(df1["PE/EA"].value_counts().head(15))

print("\nDataset 2 solvents:")
for col in ["Solvent_1", "Solvent_2", "Solvent_3"]:
    print(col, df2[col].dropna().unique())

print("\nDataset 3 solvent composition:")
print(df3[["H", "EA", "DCM", "MeOH", "Et2O"]].describe())
```
